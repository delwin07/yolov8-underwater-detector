```python
from ultralytics import YOLO

yaml_text = """path: /content/Underwater-Objects-1
train: train/images
val: valid/images
names:
  0: Control Panel
  1: Cylinder
  2: Docking Station
  3: Pipe
  4: Shell
  5: Whale
"""
open("/content/Underwater-Objects-1/data.yaml", "w").write(yaml_text)

# "before": pretrained model
YOLO("yolov8n.pt").predict("/content/Underwater-Objects-1/valid/images", save=True, conf=0.25, name="before")

# train
model = YOLO("yolov8n.pt")
model.train(data="/content/Underwater-Objects-1/data.yaml", epochs=50, imgsz=640)

# "after": fine-tuned model
best = YOLO("/content/runs/detect/train/weights/best.pt")
best.predict("/content/Underwater-Objects-1/valid/images", save=True, conf=0.4, name="after")
```
